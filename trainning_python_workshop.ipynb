# ***Bitwise Operator***

```python
a=10 #a=1010
b=5  #b=0101
print("Result=", a&b)
print("Result=", a|b)
print("Result=", a^b)
print("Result=", ~a)
print("Result=", a<<2)
print("Result=", a>>2)

a=10 # In binary: 1010
b=5  # In binary: 0101

print("Result=", a&b) # Bitwise AND: Compares each bit; if both bits are 1, the result is 1, otherwise 0. (1010 AND 0101 = 0000 = 0)
print("Result=", a|b) # Bitwise OR: Compares each bit; if either bit is 1, the result is 1, otherwise 0. (1010 OR 0101 = 1111 = 15)
print("Result=", a^b) # Bitwise XOR (Exclusive OR): Compares each bit; if the bits are different, the result is 1, otherwise 0. (1010 XOR 0101 = 1111 = 15)
print("Result=", ~a) # Bitwise NOT (One's Complement): Inverts all the bits. For positive numbers, it's equivalent to -(x + 1). (NOT 1010 = ...11110101 = -11)
print("Result=", a<<2) # Bitwise Left Shift: Shifts the bits of 'a' to the left by 2 positions, effectively multiplying 'a' by 2^2 (4). (1010 << 2 = 101000 = 40)
print("Result=", a>>2) # Bitwise Right Shift: Shifts the bits of 'a' to the right by 2 positions, effectively dividing 'a' by 2^2 (4) and discarding the remainder. (1010 >> 2 = 0010 = 2)
```
